```python
import random
print('Bienvenue dans le jeu Devinez le nombre!')
print('Je pense à un nombre entre 1 et 100. Pouvez-vous le deviner ?')
x=random.randint(1,100)
#print("Random int is : ",x)
y=0
while y!=x:
    y=int(input("Entrez un chiffre entre 1 et 100  :"))
    if y==x:
        break
    elif y<x:
        if (x-y)>20:
            print('Votre réponse est trop basse. Réessayez')
        else:
            print('Votre réponse est basse, max 20 points entre x et y. Réessayez')
    elif y>x:
        if (y-x)>20:
            print('Votre réponse est trop élevée. Réessayez')
        else:
            print('Votre réponse est élevée, max 20 points entre x et y. Réessayez')

print('Bravo')
```
